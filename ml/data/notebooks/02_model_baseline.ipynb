```python
# Imports
import pandas as pd
import numpy as np
from sklearn.metrics import mean_absolute_error
from pathlib import Path

data_path = Path("../data/processed/clean_expenses.csv")
df = pd.read_csv(data_path)
```

```python
# Mean Predictor
mean_pred = df["monthly_expense"].mean()
mae_mean = mean_absolute_error(df["monthly_expense"], [mean_pred]*len(df))
print(f"Mean Predictor MAE: {mae_mean:.2f}")
```

```python
# Seasonal Naïve (if monthly column exists)
if "month" in df.columns:
    seasonal_pred = df.groupby("month")["monthly_expense"].transform("mean")
    mae_seasonal = mean_absolute_error(df["monthly_expense"], seasonal_pred)
    print(f"Seasonal Naïve MAE: {mae_seasonal:.2f}")
```
