```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

raw_path = Path("../data/raw/student_expenses.csv")
```

```python
# Load Data
df = pd.read_csv(raw_path)
df.head()
```

```python
print(df.info())
df.describe().T
```

```python
# Missing Values
sns.barplot(x=df.isnull().mean().index, y=df.isnull().mean().values)
plt.xticks(rotation=45)
plt.title("Fraction Missing by Column")
plt.show()
```

```python
# Expense Distribution
sns.histplot(df["monthly_expense"], kde=True)
plt.title("Monthly Expense Distribution")
plt.show()
```

```python
# Category Analysis
sns.boxplot(x="category", y="amount", data=df)
plt.xticks(rotation=40)
plt.title("Amount Spent per Category")
plt.show()
```

```python
# Save Cleaned Data
df_clean = df.dropna()
df_clean.to_csv("../data/processed/clean_expenses.csv", index=False)
print("Clean dataset saved.")
```
